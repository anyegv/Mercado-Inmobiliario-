"Análisis explotratorio del Mercado Inmobilario Cubano a partir de múltiles fuentes de datos"

El mercado inmobiliario cubano es un caso singular: nació formalmente en 2011 con la legalización de la compraventa entre particulares, pero quedó atrapado casi de inmediato entre la rigidez legal, la escasez de crédito y una economía en crisis. La unificación monetaria de 2021 —que eliminó el CUC— desmontó una burbuja de precios artificiales y reveló que muchas propiedades valían menos de la mitad de lo que se anunciaban.

En los últimos años, el mercado ha sufrido un doble golpe. Por un lado, la emigración masiva puso miles de viviendas a la venta mientras los compradores con divisas se reducían drásticamente. Por otro, la inflación y el deterioro de los servicios básicos —apagones crónicos, falta de agua— hicieron que una casa en La Habana valga hoy, en promedio, entre un 40% y un 60% menos que en 2019: un apartamento en El Vedado que se pedía en 40.000 CUC hoy se ofrece por unos 22.000 USD. A esto se suma un déficit habitacional que supera las 800.000 viviendas, mientras la construcción estatal apenas levanta unas pocas miles al año. La reciente Ley 187 (2026) intenta oxigenar el sector permitiendo hasta dos propiedades y abriendo la compraventa a extranjeros, pero el acceso real sigue bloqueado para un asalariado común: el salario medio estatal ronda los 4.600 pesos mensuales, insuficientes incluso para la canasta básica.

Este proyecto parte de esa realidad y se pregunta, con datos en mano, cómo se relacionan el precio de la vivienda con sus atributos concretos. No es un análisis descriptivo más: es un intento de traducir en porcentajes y patrones verificables la intuición de que en el mercado inmobiliario habanero la ubicación, los servicios y el tamaño no suman: multiplican.

"Conclusiones a las que se ha logrado llegar con todos estos datos. Un análisis general."

Los análisis, leídos en conjunto, dibujan un mercado inmobiliario habanero que no solo es caro, sino estructuralmente disfuncional. La gráfica de servicios evidencia que el wifi, la climatización y el respaldo energético no son extras: son los verdaderos motores del sobreprecio, capaces de multiplicar por cinco el valor de un alquiler. La de cuartos muestra que el salto de una a dos habitaciones casi duplica el precio, y de dos a tres lo incrementa otro 75%: el mercado no premia los metros cuadrados, sino la habitabilidad mínima, y castiga la vivienda de un solo cuarto. La comparación internacional revela que un cubano necesita más de 80 años de salario íntegro para comprar una casa, frente a otros países latinoamericanos: una anomalía regional que no se explica por los precios, sino por salarios reales de miseria. La jerarquía por tipología confirma que el comprador paga por privacidad, estatus y potencial de negocio —no solo por ladrillos—, y la densidad por zonas demuestra que la ubicación lo es todo: Playa y Plaza de la Revolución concentran los precios más altos porque ofrecen servicios, seguridad y proyección económica, mientras Centro Habana queda atrapado en la parte baja de la distribución.

El impacto de estos resultados es doble. Hacia adentro, consolida una desigualdad espacial y social que expulsa a los hogares de menores ingresos de los municipios con mejor calidad de vida, y convierte la vivienda en un bien de lujo inalcanzable para cualquier asalariado. Hacia afuera, alimenta la emigración de jóvenes y profesionales, porque el esfuerzo de toda una vida no alcanza ni para un apartamento de un cuarto en Centro Habana. El mercado no falla por falta de demanda —el déficit supera las 800.000 viviendas—, sino por un cóctel de salarios deprimidos, crédito inexistente, construcción estatal raquítica y una regulación que, aunque empieza a abrirse con la Ley 187, sigue sin tocar el problema de fondo: el poder adquisitivo.

¿Qué podría cambiar la situación? Tres frentes concretos. Primero, política de ingresos y crédito: sin salarios reales que crezcan y sin financiamiento accesible (hipotecas indexadas, tasas subsidiadas), ninguna flexibilización legal moverá la aguja. Segundo, inversión en servicios básicos —agua, electricidad, internet— como condición para que las viviendas valgan lo que se pide por ellas, no como adorno. Tercero, construcción masiva y diversificada: permitir que cooperativas, privados y gobiernos locales construyan vivienda asequible en zonas intermedias, no solo en las ya saturadas de lujo. Y una cuarta pata, aunque incómoda: repensar el sistema de tenencia para que el alquiler social o la vivienda en usufructo con opción a compra dejen de ser tabú.

Pero más allá de las recetas, los datos invitan a una reflexión más profunda. Un país donde comprar una casa toma dos siglos de salario no tiene un mercado inmobiliario: tiene un mecanismo de expulsión silenciosa. Cada punto de la nube de servicios, cada barra de la gráfica de cuartos, cada año de la línea de asequibilidad es una historia de alguien que se queda sin futuro en su propia tierra. La pregunta ya no es solo cómo abaratar la vivienda, sino qué tipo de sociedad se está construyendo cuando el techo propio se vuelve un privilegio de pocos. Ahí está el verdadero desafío: los gráficos muestran precios, pero debajo de ellos late una pregunta por la justicia social, la productividad y el derecho a quedarse. Expandir la mente, en este caso, es entender que ningún análisis inmobiliario es neutral: cada dato sobre el precio de una casa es también un dato sobre las oportunidades —o la ausencia de ellas— de toda una generación.
